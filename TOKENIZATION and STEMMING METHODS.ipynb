```python
#                        CORPUS MEANS THE COMPLETE DATA SET.
```

```python
corpus = """Punjab Sarkar: A Look at the Mann Government

Punjab is currently governed by the Aam Aadmi Party (AAP), led by Chief Minister Bhagwant Mann, who took office on 16 March 2022. AAP's win was historic. It won around three-fourths of the assembly seats and broke the long-standing dominance of Congress and the Shiromani Akali Dal in the state.

The government has focused on a few flagship areas. Aam Aadmi Clinics were set up to bring basic healthcare closer to villages and towns. Free electricity up to 300 units per month has eased household bills. The state has also pushed upgrades in government schools, along with anti-corruption drives and campaigns against drug abuse, one of Punjab's most serious social problems.

Recently, the government set up a new State Finance Commission to strengthen the finances of urban and rural local bodies, aiming to improve grassroots governance.

Challenges remain. Punjab carries a heavy debt burden, and farmers continue to face issues around crop diversification, groundwater depletion, and stubble burning. Youth unemployment and migration abroad are also major concerns. Critics question whether welfare schemes are financially sustainable, while supporters point to visible improvements in services.

With the next assembly elections due in early 2027, the coming months will decide how voters judge this government's record."""

print(corpus)
```

```python
#                       FIRST WAY TO TOKENIZE FROM PARAGAPGH TO SENTENCES.
```

```python
from nltk.tokenize import sent_tokenize
```

```python
import nltk
nltk.pathsec.ALLOW_PROXIED_FETCH = True
nltk.download('punkt_tab')
```

```python
sentences=sent_tokenize(corpus)
print(sentences)
```

```python
sentences=sent_tokenize(corpus)
print(len(sentences))
```

```python
#                            TO CONVERT THE SENTENCES INTO LIST
```

```python
for i in sentences:
  print(i)
```

```python
#                     NOW CONVERTING SENTENCES TO TOKENS (WORDS)
```

```python
from nltk.tokenize import word_tokenize
```

```python
words=word_tokenize(corpus)
print(words)
```

```python
#                    SAM WAY TO MAKE THE RESULT IN WORDS TOKENS
```

```python
words=word_tokenize(sentences[0])
print(words)
```

```python
#                     PRINT THE SENTENCES IN LINE
```

```python
for i in sentences:
  words=word_tokenize(i)
  print(words,'---->',len(words))
```

```python
#                         WORD PUCT TOKENIZATION
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
for i in sentences:
  words=wordpunct_tokenize(i)
  print(words,'---->',len(words))
```

```python
#                      TREE BANK WORDS TOKENIZATION METHOD
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
#                         FIRST INITIALIZE THE TOKENS
```

```python
tokens=TreebankWordTokenizer()
```

```python
for i in sentences:
  tokens=TreebankWordTokenizer()
  words=tokens.tokenize(i)
  print(words,'---->',len(words))
```

```python
#                          SPLIT METHOD TOKENIZATION
```

```python
for i in sentences:
  words=i.split()
  print(words,'---->',len(words))
```

```python
#                                   STEMMING
```

```python
from nltk.stem import PorterStemmer
```

```python
words=['playing','Games','Scrolling','Internet']
```

```python
#                   FIRST INITIALIZE THE STEMMER
```

```python
stemmer=PorterStemmer()
```

```python
for i in words:
  print(stemmer.stem(i))
```

```python
sentences[0]
```

```python
for i in sentences[0].split():
  print(i,'--->',stemmer.stem(i))
```

```python
for i in sentences:
  words=word_tokenize(i)
  for j in words:
    print(stemmer.stem(j))
```

```python
#                                  SNOW BALL METHOD STEMMING
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowball=SnowballStemmer('english')
```

```python
for word in words:
  print(words,'---->',snowball.stem(word))
```

```python
for i in sentences[0].split():
  print(i,'---->',snowball.stem(i))
```
